```python
import pandas as pd
import os 
```

```python
env = pd.read_csv("../data/raw/env_air_.csv")
sdg = pd.read_csv("../data/raw/sdg_07_40.csv")
```

### Modelação do DF: SDG

```python
sdg.head()
```

```python
sdg_long = sdg.copy()
sdg_long = sdg_long.melt(id_vars=["freq", "nrg_bal", "unit", "geo\\TIME_PERIOD"], var_name="year", value_name="percentagem")
sdg_long.head()
```

```python
columns = ["freq", "nrg_bal", "unit"]

for cols in columns:
    if cols not in columns:
        continue
    check = sdg_long[cols].unique()
    if len(check) == 1:
        sdg_long = sdg_long.drop(columns=cols)
        print("Coluna removida:", cols)
    else:
        print(cols, "->", check)
```

```python
sdg_long = sdg_long.rename(columns={
                           	"nrg_bal": "energia_ren",
                            "geo\TIME_PERIOD": "pais"})

sdg_long.head()
```

### Modelação do DF: ENV

```python
env.head()
```

```python
env_long = env.copy()
env_long = env_long.melt(id_vars=["freq", "unit", "airpol", "src_crf", "geo\\TIME_PERIOD"], var_name="year", value_name="mio_t_co2eq")
env_long.head()
```

```python
columns = ["freq", "unit", "airpol", "src_crf"]

for cols in columns:
    if cols not in env_long.columns:
        continue
    check = env_long[cols].unique()
    if len(check) == 1:
        env_long = env_long.drop(columns=cols)
        print("Coluna removida:", cols)
    else:
        print(cols, "->", check)
```

```python
env_long = env_long.rename(columns={
                            "geo\TIME_PERIOD": "pais"})
```

```python
env_long = env_long[env_long["year"]>="2004"]
env_long
```

```python
env_long[env_long["airpol"]=="GHG"]["unit"].unique()
```

```python
ghg = env_long[env_long["airpol"]=="GHG"]
paises_mio = set(ghg[ghg["unit"] == "MIO_T"]["pais"].unique())
paises_ths = set(ghg[ghg["unit"] == "THS_T"]["pais"].unique())

print("Só em MIO_T:", paises_mio - paises_ths)
print("Só em THS_T:", paises_ths - paises_mio)
print("Em ambos:", paises_mio & paises_ths)
```

```python
env_long = env_long[env_long["unit"] == "MIO_T"]
env_long = env_long[env_long["airpol"]=="GHG"]
```

```python
paises_com_total = set(env_long[env_long["src_crf"] == "TOTXMEMO"]["pais"].unique())
todos_os_paises = set(env_long["pais"].unique())

print("Sem TOTXMEMO:", todos_os_paises - paises_com_total)
```

```python
env_long = env_long[env_long["src_crf"] == "TOTXMEMO"]
print(env_long.head())
```

### Criação do Dataset a ser trabalhado

```python
sdg_long["year"].dtype
env_long["year"].dtype
```

```python
merged_eu = pd.merge(sdg_long, env_long, on=["pais", "year"], how="inner")
merged_eu = merged_eu.drop(columns=["unit", "airpol", "src_crf"], errors="ignore")
print(merged_eu.head())
```

```python
eu27 = ["AT", "BE", "BG", "HR", "CY", "CZ", "DK", "EE", "FI", "FR",
        "DE", "EL", "HU", "IE", "IT", "LV", "LT", "LU", "MT", "NL",
        "PL", "PT", "RO", "SK", "SI", "ES", "SE"]

merged_eu_27 = merged_eu[merged_eu["pais"].isin(eu27).copy()]

paises_encontrados = set(merged_eu["pais"].unique())
paises_27 = set(merged_eu_27["pais"].unique())

print("Em falta:", paises_encontrados - paises_27)
print("Em falta:", set(eu27) - paises_encontrados)
```

```python
merged_eu_27["meta_eu"] = 42.5 - merged_eu_27["percentagem"]
merged_eu_27
```

```python
merged_eu_27.to_csv("../data/raw/processed/eu_renewable_emissions.csv", index=False)
```
